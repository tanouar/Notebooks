### Business case

You serve as Chief Systems Architect for the Rebel Alliance. The movement operates across the galaxy with pilots, starships, and missions that must coordinate. 
Your current approach uses separate functions for each task with disconnected data structures. 

### The procedural approach breaks at scale

```python
# Procedural approach: separate data structures require manual coordination
pilot_names = []
pilot_callsigns = []
pilot_experience = []
pilot_missions = []

def create_pilot(name, callsign, experience=1):
    """Add new pilot across four parallel lists"""
    pilot_names.append(name)
    pilot_callsigns.append(callsign)
    pilot_experience.append(experience)
    pilot_missions.append(0)
    return len(pilot_names) - 1

def complete_mission(pilot_index):
    """Update two separate lists for mission completion"""
    pilot_missions[pilot_index] += 1
    pilot_experience[pilot_index] += 2
```

```python
# Create pilots and track indices manually
luke_idx = create_pilot("Luke Skywalker", "Red Five", 3)
wedge_idx = create_pilot("Wedge Antilles", "Red Two", 5)

complete_mission(luke_idx)
complete_mission(luke_idx)

print(f"{pilot_callsigns[luke_idx]}: {pilot_missions[luke_idx]} missions, Level {pilot_experience[luke_idx]}")
```

This code demonstrates **three critical failures**: 
- Manual index tracking creates fragility. 
- Lose track of these numbers and you update the wrong pilot's data. 
- Parallel list synchronization has no enforcement mechanism. Nothing prevents appending to `pilot_names` without updating `pilot_callsigns`. Adding new attributes cascades changes everywhere. The approach cannot scale.

**Object-Oriented Programming** solves this by bundling related data with the functions that operate on that data. Instead of coordinating separate lists, you create Pilot objects that contain all their own information and know how to manage themselves.

## Classes and objects

A class defines **structure** and **behavior** as a blueprint. 

An object creates an actual **instance** from that blueprint. 

<img src = "https://data-analytics-fullstack-assets.s3.eu-west-3.amazonaws.com/M05-Python_Programming/M2_D3_Class_object.png"/>

When you write `class Pilot`, you define what all pilots share. When you write `luke = Pilot("Luke Skywalker", "Red Five")`, you create an actual pilot with its own data.

Now let’s take a look at the code.

```python
class Pilot:
    """
    Manages a Rebel pilot with missions and experience
    
    This is a CLASS DEFINITION - a blueprint for creating pilot objects.
    Think of it like an architectural plan: it describes what every pilot
    will have (attributes) and what every pilot can do (methods), but it
    doesn't create any actual pilots yet.
    """
    
    def __init__(self, name, callsign, experience=1):
        """
        Initialize pilot with required parameters
        
        THE CONSTRUCTOR: This special method runs automatically when you
        create a new pilot object. Python calls it behind the scenes.
        
        Parameters:
            self: The object being created (Python passes this automatically)
            name: Pilot's full name (required)
            callsign: Pilot's identification (required)
            experience: Starting skill level (optional, defaults to 1)
        
        When you write: luke = Pilot("Luke Skywalker", "Red Five", 3)
        Python actually does: Pilot.__init__(luke, "Luke Skywalker", "Red Five", 3)
        """
        # INSTANCE ATTRIBUTES: These belong to THIS specific pilot object
        # Each pilot gets their own separate copy of these values
        self.name = name              # Store the name on this pilot object
        self.callsign = callsign      # Store the callsign on this pilot object
        self.experience = experience  # Store experience level on this pilot object
        self.missions = 0             # Initialize mission counter to zero
        
        # The "self." prefix means "attach this to the object being created"
        # Without "self.", these would be local variables that disappear
        # when __init__ finishes running
    
    
    def complete_mission(self):
        """
        Update this pilot's mission history
        
        INSTANCE METHOD: A function that operates on a specific pilot object.
        
        Parameters:
            self: The pilot object calling this method (passed automatically)
        
        When you write: luke.complete_mission()
        Python actually does: Pilot.complete_mission(luke)
        """
        # Access THIS pilot's attributes using self
        # self.missions refers to the missions count for whichever pilot
        # called this method (could be luke.missions, wedge.missions, etc.)
        self.missions += 1      # Increment this pilot's mission count by 1
        self.experience += 2    # Increase this pilot's experience by 2 points
        
        # Return a formatted string using this pilot's callsign and mission count
        # The f-string pulls values from self.callsign and self.missions
        return f"{self.callsign} completed mission #{self.missions}"
    
    def get_summary(self):
        """
        Generate pilot status summary
        
        INSTANCE METHOD: Returns formatted information about this pilot.
        
        Parameters:
            self: The pilot object calling this method
        
        Returns:
            A string containing this pilot's key information
        """
        # Build a summary string using this pilot's attributes
        # Each pilot returns their own unique values when this method runs
        return f"{self.callsign} ({self.name}): {self.missions} missions, Level {self.experience}"
```

## Attributes

**Instance attributes** belong to individual objects. When you write `self.missions = 0` inside `__init__`, you create an attribute that belongs to that specific object. Each instance gets its own copy. 

**Class attributes** are shared across all instances. When you write `minimum_age = 18` directly in the class body, you create an attribute shared by all pilots.

```python
class Pilot:
    """Pilot with instance and class attributes"""
    
    # Class attributes: shared across all pilots
    minimum_age = 18
    minimum_experience = 1
    total_pilots = 0
    
    def __init__(self, name, callsign, experience=1):
        """Initialize instance attributes"""
        # Instance attributes: unique to this pilot
        self.name = name
        self.callsign = callsign
        self.experience = experience
        self.missions = 0
        
        # Update class-level counter
        Pilot.total_pilots += 1
    
    def complete_mission(self):
        self.missions += 1
        self.experience += 2
```

## Encapsulation

Direct attribute access creates data corruption. When any code can write `pilot.missions = -5` or `pilot.experience = "banana"`, your program accepts nonsense. The pilot object cannot defend its own integrity.

Encapsulation transfers control from callers to the object itself. The object declares that external code cannot touch internal data directly. Every change must pass through methods or properties that enforce rules.

**Python uses the underscore prefix to signal private attributes.** Writing `_missions` tells other developers this attribute belongs to the class implementation, not the public interface. 

<Note type="important">

The underscore creates a convention, not a lock. Python trusts developers to respect boundaries rather than enforcing them through compiler errors.

</Note>

Properties provide the enforcement mechanism. When you write `pilot.experience`, Python routes that access through a property method. That method runs validation code before allowing the change. Invalid values get rejected. The syntax looks like simple attribute access, but the property intercepts every operation.

Consider this unprotected implementation:

```python
class Pilot:
    def __init__(self, name):
        self.name = name
        self.missions = 0
        self.experience = 0
```

Any code can now corrupt the pilot's state:

```python
pilot = Pilot("Jenkins")
pilot.missions = -100  # Accepts impossible value
pilot.experience = "expert"  # Accepts wrong type
pilot.random_attr = 42  # Creates attributes that shouldn't exist
```

Anyone can modify any attribute with any value. No validation exists. Typos create new attributes silently. This breaks in production when negative missions corrupt payroll or zero experience pilots get assigned to dangerous missions.

**Encapsulation solves this by hiding internal data and exposing only controlled methods:**


```python
class Pilot:
    def __init__(self, name, callsign):
        self.name = name
        self._missions = 0        # Underscore signals "private - don't touch"
        self._experience = 1
    
    def complete_mission(self):
        self._missions += 1       # Only increments - never decrements
        self._experience += 2     # Maintains valid relationship
        return f"Mission #{self._missions} complete"
    
    @property
    def missions(self):           # Read-only access through property
        return self._missions
    
    @property
    def experience(self):
        return self._experience
```

```python
luke = Pilot("Luke Skywalker", "Red Five")
luke.complete_mission()
print(luke.missions)  # Output: 1

try:
    luke.missions = -50
except AttributeError as e:
    print(f"Assignment blocked: {e}")
    # Output: Assignment blocked: can't set attribute
```

The `@property` decorator transforms the `missions` method into a read-only attribute. External code can call `luke.missions` to retrieve the value, but any attempt to assign `luke.missions = -50` triggers an AttributeError. Python refuses the operation because no setter method exists.

The `complete_mission` method provides the only valid path to change mission count. This method accesses `_missions` directly because it lives inside the class. The method enforces business rules: missions only increment by one, they never decrease, and they never accept arbitrary values from outside.

This pattern prevents three corruption scenarios. First, negative mission counts become impossible because `complete_mission` only adds positive increments. Second, type errors disappear because the method controls all assignments. Third, the mission count always reflects actual completed missions rather than random values injected from external code.

The object now owns its data integrity. External code cannot bypass validation rules or create inconsistent state. Every mission count you read accurately represents missions that passed through `complete_mission`.

<Note type="tip">

A setter lets you control writes to a property while still using assignment syntax. You add validation logic that runs before Python stores the value. When you write `pilot.experience = 150`, Python calls your setter method, which checks whether 150 passes validation rules before updating the private attribute.

Without a setter, the property becomes read-only. With a setter, you maintain the natural assignment syntax while enforcing constraints that protect data integrity.

</Note>

```python
class Pilot:
    def __init__(self, name, callsign):
        self.name = name
        self.callsign = callsign
        self._missions = 0
        self._experience = 0
    
    @property
    def experience(self):
        return self._experience
    
    
    @experience.setter
    def experience(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Experience must be a number")
        if value < 0:
            raise ValueError("Experience cannot be negative")
        self._experience = value
```

The setter method runs every time external code assigns to `experience`. The validation checks type and range before updating `_experience`. Invalid values trigger exceptions that prevent corruption:

```python
luke = Pilot("Luke Skywalker", "Red Five")
luke.experience = 150  # Works - passes validation
```

```python
try:
    luke.experience= "expert"
except TypeError as e:
    print(f"Assignment blocked: {e}")
    # Output: Assignment blocked: can't set attribute
```

## Method types

In Python, methods come in **three main flavors** depending on what they act on: the individual object, the class itself, or neither (just a related utility function).


- Instance methods receive `self` and operate on **one specific object**. They can read and modify that object’s state: `pilot.complete_mission()` updates *this* pilot only.

- Class methods receive `cls` and operate on the **class as a whole**. They’re often used for alternative constructors or for actions that concern all instances, e.g. `Pilot.from_dict(...)` or `Pilot.total_pilots`.

- Static methods receive neither `self` nor `cls`. They’re **utility functions** that are related to the class conceptually, but don’t need access to instance or class state, e.g. `Pilot.validate_callsign(callsign)`. They live inside the class to keep related logic grouped in one place.


```python
class Pilot:
    """Pilot with different method types"""
    
    total_pilots = 0
    pilot_registry = {}
    
    def __init__(self, name, callsign, experience=1):
        self.name = name
        self.callsign = callsign
        self.experience = experience
        self.missions = 0
        
        # Register pilot
        Pilot.total_pilots += 1
        Pilot.pilot_registry[callsign] = self
    
    # Instance method - operates on specific pilot
    def complete_mission(self):
        """Update this pilot's stats"""
        self.missions += 1
        self.experience += 2
        return f"{self.callsign}: Mission #{self.missions} complete"
    
    # Class method - operates on class-level data
    @classmethod
    def get_pilot_count(cls):
        """Return total registered pilots"""
        return cls.total_pilots
    
    # Class method - alternative constructor
    @classmethod
    def find_pilot(cls, callsign):
        """Look up pilot by callsign"""
        return cls.pilot_registry.get(callsign)
    
    # Static method - utility function
    @staticmethod
    def validate_callsign(callsign):
        """Check if callsign follows naming convention"""
        valid_prefixes = ['Red', 'Blue', 'Gold', 'Green']
        return any(callsign.startswith(prefix) for prefix in valid_prefixes)
```

```python
# Create pilots
luke = Pilot("Luke Skywalker", "Red Five", 3)
wedge = Pilot("Wedge Antilles", "Red Two", 5)
```

```python
# Instance method works with specific pilot
print(luke.complete_mission())
```

```python
# Class method works with class-level data
print(f"\nTotal pilots: {Pilot.get_pilot_count()}")

# Class method looks up from registry
found = Pilot.find_pilot("Red Five")
print(f"Found: {found.name if found else 'Not found'}")
```

```python
# Static method validates without instance or class state
print(f"'Red Five' valid: {Pilot.validate_callsign('Red Five')}")
print(f"'Maverick' valid: {Pilot.validate_callsign('Maverick')}")
```

## Special methods

Special methods integrate objects with built-in operations. Define `__str__` to control string conversion. Define `__eq__` to support equality comparison. Define `__lt__` to enable sorting.

| Method | Controls | Example Usage |
|--------|----------|---------------|
| `__str__` | String representation for humans | `print(pilot)` calls `pilot.__str__()` |
| `__repr__` | String representation for developers | Console display and debugging output |
| `__eq__` | Equality comparison | `pilot1 == pilot2` calls `pilot1.__eq__(pilot2)` |
| `__lt__` | Less-than comparison | `pilot1 < pilot2` enables sorting operations |
| `__len__` | Length calculation | `len(squadron)` returns squadron size |
| `__add__` | Addition operator | `fleet1 + fleet2` combines two fleets |
| `__getitem__` | Index access | `squadron[0]` retrieves first pilot |
| `__iter__` | Iteration protocol | `for pilot in squadron` loops through members |

Python calls these methods automatically when your objects encounter built-in operations. You write the method once, and Python integrates your object with the language's standard syntax. This transforms custom objects into first-class citizens that behave like built-in types.


```python
class Pilot:
    """Pilot with special methods for natural operations"""
    
    def __init__(self, name, callsign, experience=1):
        self.name = name
        self.callsign = callsign
        self.experience = experience
        self.missions = 0
        self.kills = 0
    
    def complete_mission(self, kills=0):
        self.missions += 1
        self.kills += kills
        self.experience = min(10, self.experience + 1)
    
    def __str__(self):
        """User-friendly string representation"""
        return f"{self.callsign} ({self.name}) - Level {self.experience}"
    
    def __repr__(self):
        """Developer-focused representation"""
        return f"Pilot('{self.name}', '{self.callsign}', {self.experience})"
    
    def __eq__(self, other):
        """Equality based on callsign"""
        if not isinstance(other, Pilot):
            return False
        return self.callsign == other.callsign
    
    def __lt__(self, other):
        """Compare by experience then kills"""
        if self.experience != other.experience:
            return self.experience < other.experience
        return self.kills < other.kills
    
    def __hash__(self):
        """Make pilots hashable for use in sets/dicts"""
        return hash(self.callsign)
```

```python
# Create pilots
luke = Pilot("Luke Skywalker", "Red Five", 4)
wedge = Pilot("Wedge Antilles", "Red Two", 6)
biggs = Pilot("Biggs Darklighter", "Red Three", 5)

luke.complete_mission(kills=3)
wedge.complete_mission(kills=2)
```

```python
# __str__ for user display
print("User view:")
print(f"  {luke}")
```

```python
# __repr__ for development
print(f"\nDeveloper view: {repr(wedge)}")
```

```python
# __eq__ for equality
luke2 = Pilot("Luke Skywalker", "Red Five", 4)
print(f"\nluke == luke2: {luke == luke2}")
print(f"luke == wedge: {luke == wedge}")
```

```python
# __lt__ enables sorting
pilots = [luke, wedge, biggs]
pilots.sort(reverse=True)
print("\nPilots ranked by experience:")
for i, pilot in enumerate(pilots, 1):
    print(f"  {i}. {pilot}")
```

## Complete pilot management system

This complete example integrates all fundamental OOP concepts into a working system.

```python
class Pilot:
    """Complete pilot management with all OOP fundamentals"""
    
    # Class attributes
    minimum_experience = 1
    total_pilots = 0
    pilot_registry = {}
    
    def __init__(self, name, callsign, experience=1):
        # Validation
        if not name or not callsign:
            raise ValueError("Name and callsign required")
        if experience < self.minimum_experience:
            raise ValueError(f"Experience must be at least {self.minimum_experience}")
        
        # Instance attributes
        self.name = name
        self.callsign = callsign
        self._experience = experience
        self._missions = 0
        self._kills = 0
        self.status = 'Ready'
        
        # Register
        Pilot.total_pilots += 1
        Pilot.pilot_registry[callsign] = self
    
    # Properties for read-only access
    @property
    def experience(self):
        return self._experience
    
    @property
    def missions(self):
        return self._missions
    
    @property
    def kills(self):
        return self._kills
    
    @property
    def is_ace(self):
        return self._kills >= 5
    
    @property
    def success_rate(self):
        if self._missions == 0:
            return 0.0
        return (self._kills / self._missions) * 100
    
    # Instance methods
    def complete_mission(self, kills=0):
        if kills < 0:
            raise ValueError("Kills cannot be negative")
        
        self._missions += 1
        self._kills += kills
        self._experience = min(10, self._experience + 1)
        return f"{self.callsign}: Mission #{self._missions} complete"
    
    def get_summary(self):
        rank = "Ace" if self.is_ace else "Pilot"
        return f"""
{self.callsign} - {self.name} ({rank})
  Experience: Level {self._experience}
  Missions: {self._missions}
  Kills: {self._kills}
  Success Rate: {self.success_rate:.1f}%
  Status: {self.status}
        """
    
    # Class methods
    @classmethod
    def get_pilot_count(cls):
        return cls.total_pilots
    
    @classmethod
    def find_pilot(cls, callsign):
        return cls.pilot_registry.get(callsign)
    
    @classmethod
    def list_all_pilots(cls):
        return list(cls.pilot_registry.keys())
    
    # Static method
    @staticmethod
    def validate_callsign(callsign):
        valid_prefixes = ['Red', 'Blue', 'Gold', 'Green']
        return any(callsign.startswith(prefix) for prefix in valid_prefixes)
    
    # Special methods
    def __str__(self):
        return f"{self.callsign} ({self.name}) - Level {self._experience}"
    
    def __repr__(self):
        return f"Pilot('{self.name}', '{self.callsign}', {self._experience})"
    
    def __eq__(self, other):
        if not isinstance(other, Pilot):
            return False
        return self.callsign == other.callsign
```

```python
# Create pilot roster
luke = Pilot("Luke Skywalker", "Red Five", 3)
wedge = Pilot("Wedge Antilles", "Red Two", 5)
biggs = Pilot("Biggs Darklighter", "Red Three", 4)
Tarik = Pilot("Tarik", "Green One", 6)
```

```python
# Display summaries
print("\n=== PILOT SUMMARIES ===")
print(luke.get_summary())
print(wedge.get_summary())
```

```python
# Use class methods
print("=== FLEET STATUS ===")
print(f"Total pilots: {Pilot.get_pilot_count()}")
print(f"Registered: {', '.join(Pilot.list_all_pilots())}")
```

```python
# Lookup pilot
found = Pilot.find_pilot("Red Five")
print(f"\nFound pilot: {found}")
```

```python
# Validate callsign
print(f"\nCallsign validation:")
print(f"  'Red Five': {Pilot.validate_callsign('Red Five')}")
print(f"  'Maverick': {Pilot.validate_callsign('Maverick')}")
```

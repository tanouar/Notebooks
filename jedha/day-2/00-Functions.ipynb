## Why functions matter
In programming, we follow the **DRY principle** - "**Don't Repeat Yourself**". This fundamental concept means that every piece of knowledge or logic should have a **single**, **unambiguous representation** in your code. When you find yourself **copying and pasting** code, it's time to create a **function**.

**Functions** provide several technical benefits:
- **Code reusability**: Write once, call multiple times
- **Modularity**: Break complex problems into smaller, manageable pieces
- **Easier debugging**: Test and fix issues in one location
- **Improved readability**: Give meaningful names to blocks of logic
- **Memory efficiency**: Code is stored once in memory

Imagine calculating hyperspace jump distances between star systems. Without **functions**, you'd write the same complex calculation for every single route - Tatooine to Hoth, Hoth to Dagobah, Dagobah to Endor. That's hundreds of **repeated calculations**, each a potential source of errors. When the Empire changes hyperspace lane regulations (and they will), you update **one** **function** instead of hunting through thousands of lines of code.

## Creating your first function
A **function** in Python is defined using the **`def`** **keyword**, followed by a **function** **name**, **parentheses** containing **parameters**, and a **colon**. The **function** body must be **indented** (typically 4 spaces) and contains the code to be executed when the **function** is **called**.

**Functions** are structured, they have a clear name, accept specific inputs (**arguments**), execute a series of **instructions**, and optionally **return** results. Here's the anatomy:

```python
def function_name(arguments):
    # Instructions to execute
    return result  
```



Let's create our first **function** for calculating the power output of a starship's reactor:



```python
def calculate_reactor_power(base_power, efficiency):
    """Calculate actual power output based on reactor efficiency"""
    actual_power = base_power * (efficiency / 100)
    print(f"Reactor generating {actual_power} gigawatts")
    return actual_power

# Mission execution - calling the function
x_wing_power = calculate_reactor_power(1000, 95)
tie_fighter_power = calculate_reactor_power(800, 88)
millennium_falcon = calculate_reactor_power(2000, 75) 
```

Notice how we write the calculation **once** but use it for different starships. This is the utility of **functions**, **write once, use everywhere**.

## Docstrings
A **docstring** is a **string literal** that occurs as the first statement in a **function**, module, class, or method definition. It becomes the **`__doc__` special attribute** of that object. **Docstrings** serve as **inline documentation** that can be accessed programmatically.

Just as every Rebel mission needs a clear briefing, every **function** needs documentation. **Docstrings** are your **function**'s technical manual. It is essential for when other developers (or **future you**) need to understand or maintain the code.

### Writing effective docstrings
**Docstrings** use **triple quotes** (""") and follow specific conventions. The **PEP 257** convention recommends starting **with** a **one-line summary**, followed by a blank line and more elaborate description if needed.



```python
def calculate_shield_strength(base_power, efficiency_percentage, damage_taken=0):
    """
    Calculate remaining shield strength after damage.
    
    This function determines the actual shield strength of a spacecraft
    based on generator power, efficiency, and any damage sustained.
    
    Args:
        base_power (int): Base shield generator power in megawatts
        efficiency_percentage (float): Shield efficiency as percentage (0-100)
        damage_taken (int, optional): Damage already sustained. Defaults to 0.
    
    Returns:
        float: Actual shield strength in megawatts
        
    Example:
        >>> calculate_shield_strength(1000, 85.5, 200)
        655.0
    """
    actual_shields = (base_power * efficiency_percentage / 100) - damage_taken
    return max(0, actual_shields)  # Shields can't go negative
```

## The Return statement 
The **`return`** **statement** exits a **function** and optionally passes data back to the caller. Without **`return`**, **functions** perform actions but don't provide values that can be used elsewhere in your program. Without it, your calculations stay **trapped** inside the  **function**, inaccessible to the rest of your program.



```python
def analyze_imperial_strength(troops, star_destroyers):
    """Calculate total Imperial force strength"""
    # Each Star Destroyer carries 37,000 troops
    total_force = troops + (star_destroyers * 37000)
    threat_level = "High" if total_force > 100000 else "Moderate"
    
    # Without return, this data dies with the function
    return total_force, threat_level

# Retrieve and use the intelligence
force_size, threat = analyze_imperial_strength(50000, 3)
print(f"Imperial force: {force_size} troops - Threat: {threat}")
```



<Note type="note">
    
The **`return`** statement **immediately exits** the **function**. Any code after **`return`** won't execute.
</Note>

## Multiple Arguments
**Functions** can accept **multiple** **parameters** (also called **arguments**). **Parameters** are defined in the **function** **signature** and act as variables within the **function** **scope**. When calling a **function**, you pass **arguments** that get assigned to these **parameters**.

Real missions require multiple inputs. When calculating hyperspace routes, you need starting coordinates, destination, fuel capacity, and cargo weight. **Functions** elegantly handle multiple **arguments**:



```python
a, b = 5, 6

print(a)
print(b)
```

```python
def calculate_hyperspace_jump(origin, destination, ship_class, cargo_weight=0):
    """Calculate jump distance and fuel consumption"""
    # Base distances (in parsecs)
    routes = {
        ("Tatooine", "Hoth"): 7500,
        ("Hoth", "Dagobah"): 4200,
        ("Dagobah", "Endor"): 5800
    }
    
    # Ship fuel efficiency (parsecs per unit)
    efficiency = {
        "X-Wing": 100,
        "Millennium Falcon": 85,
        "Star Destroyer": 50
    }
    
    distance = routes.get((origin, destination), 10000)  # Default unknown route
    base_fuel = distance / efficiency.get(ship_class, 75)
    
    # Cargo adds 10% fuel consumption per 1000 tons
    cargo_penalty = 1 + (cargo_weight / 10000)
    total_fuel = base_fuel * cargo_penalty

    return distance, total_fuel # fin de la fonction



# Different ways to call the function
distance, fuel = calculate_hyperspace_jump("Tatooine", "Hoth", "X-Wing")
distance2, fuel2 = calculate_hyperspace_jump("Hoth", "Dagobah", "Millennium Falcon", 500)

# Using named arguments for clarity (order doesn't matter)
distance3, fuel3 = calculate_hyperspace_jump(
    destination="Endor",
    origin="Dagobah", 
    ship_class="Star Destroyer",
    cargo_weight=10000
)
```



## Default arguments 
**Default** **arguments** allow **parameters** to have **predetermined values** if not explicitly provided during the **function** call. They must be defined after non-default **parameters** in the **function** signature.

Sometimes you have standard values that rarely change. Default **arguments** let you define these once:



```python
def scan_for_life_forms(planet, scan_depth=100, include_droids=False):
    """Scan planet for life forms with configurable parameters"""
    print(f"Scanning {planet} to depth of {scan_depth} meters")
    
    life_forms = {
        "Tatooine": 85000000,
        "Hoth": 12000,
        "Dagobah": 95000000000  # So much life...
    }
    
    count = life_forms.get(planet, 0)
    if include_droids:
        count += 500000  # Rough estimate of droid population
    
    return count

# Use defaults for standard scan
tatooine_life = scan_for_life_forms("Tatooine", 600)

# Override defaults for deep scan including droids
detailed_scan = scan_for_life_forms("Tatooine", scan_depth=500, include_droids=True)
```



<Note type="warning">
    
Default **arguments** must come **after** non-default **arguments**. Python reads left to right, so `def function(default=10, required)` will cause an error.
</Note>

## Managing exceptions 
**Exceptions** are events that occur during program execution that disrupt the normal flow of instructions. Python provides a structured way to handle these **exceptions** using **`try`**, **`except`**, **`else`**, and **`finally`** blocks.

Not every mission goes as planned. The Death Star had a thermal exhaust port, the hyperdrive can malfunction, and users might input invalid data. **Exception** handling prevents your program from crashing like an X-Wing into a swamp.

### Understanding exceptions
When Python encounters an error, it creates an **exception** **object**. If not handled, the program **terminates** and prints a **traceback**. Common **exception** types include **`ValueError`**, **`TypeError`**, **`ZeroDivisionError`**, and **`KeyError`**.

Let's see what happens when things go wrong:



```python
def calculate_fleet_strength(ships, crew_per_ship):
    """Calculate total fleet personnel"""
    total_crew = ships * crew_per_ship
    return total_crew

# This works fine
rebels = calculate_fleet_strength(30, 5)

# But what if someone enters text instead of numbers?
# empire = calculate_fleet_strength("many", 5)  # TypeError!
```



Without **exception** handling, **one bad input** crashes your entire intelligence system.

### `Try`/`Except` 
The **`try`** block contains code that might **raise** an **exception**. The **`except`** block catches and handles the **exception**. The **`else` block** runs if no **exception** occurs, and **`finally`** always executes regardless of **exceptions**.


```python
def process_imperial_transmission(encrypted_data, decryption_key):
    """Decrypt intercepted Imperial communications"""
    try:
        # Attempt decryption (risky operation)
        decrypted = int(encrypted_data) / decryption_key
        message = f"Fleet size: {decrypted} ships"
    except ValueError:
        # Bad data format - not a number
        message = "Transmission corrupted - cannot parse data"
    except ZeroDivisionError:
        # Invalid decryption key
        message = "Invalid decryption key - cannot decode"
    except Exception as e:
        # Catch any other unexpected errors
        message = f"Unknown error in decryption: {e}"
    else:
        # Only runs if no exceptions occurred
        print("Decryption successful")
    
    return message

# Test different scenarios
print(process_imperial_transmission("1000", 10))    # Success
print(process_imperial_transmission("XXX", 10))     # Corrupted
print(process_imperial_transmission("1000", 0))     # Bad key
```



### Multiple exception types 
You can handle different **exception** types **separately**, providing specific responses for each error condition. This allows for **precise error handling** and better user feedback.

```python
def launch_attack_run(target, torpedoes, distance):
    """Simulate Death Star attack run"""
    try:
        # Check torpedo count
        if torpedoes < 0:
            raise ValueError("Torpedo count cannot be negative")
        
        # Calculate hit probability
        accuracy = 100 / distance  # Closer = better
        
        # Need at least 2 torpedoes for the shot
        if torpedoes < 2:
            raise RuntimeError("Insufficient torpedoes for attack run")
            
        success_chance = accuracy * torpedoes * 10
        
    except ValueError as e:
        print(f"Invalid input: {e}")
        return 0
    except ZeroDivisionError:
        print("Target too far - cannot calculate trajectory")
        return 0
    except RuntimeError as e:
        print(f"Mission abort: {e}")
        return 0
    else:
        print(f"Attack run viable - {success_chance}% success chance")
        return success_chance

# Different failure modes
launch_attack_run("Death Star", 2, 10)    # Success
launch_attack_run("Death Star", 1, 10)    # Not enough torpedoes
launch_attack_run("Death Star", -5, 10)   # Invalid input
launch_attack_run("Death Star", 2, distance=0)     # Too close
```



### Creating custom exceptions
You can **raise** **exceptions** manually using the **`raise`** statement. This is useful for enforcing **business rules** or **data validation** that Python wouldn't naturally check.

Sometimes standard **exceptions** aren't specific enough. You can create your own for special situations:



```python
def manage_rebel_resources(credits, fuel, food_supplies):
    """Allocate resources for rebel base operations"""
    try:
        # Convert inputs to numbers
        credits = float(credits)
        fuel = float(fuel)
        food = float(food_supplies)
        
        # Check for negative values (data corruption?)
        if credits < 0 or fuel < 0 or food < 0:
            raise ValueError("Resources cannot be negative - possible Imperial tampering")
        
        # Check critical thresholds
        if fuel < 1000:
            raise Exception("CRITICAL: Fuel reserves below minimum for evacuation")
        
        if food < 30:
            raise Exception("WARNING: Food supplies critical - less than 30 days remaining")
        
        # Calculate resource score
        resource_health = (credits/10000 + fuel/5000 + food/100) / 3
        
    except ValueError as e:
        print(f"Data validation failed: {e}")
        return None
    except Exception as p:
        print(f"⚠️ {p}")
        # Still calculate but flag the warning
        resource_health = -1
    else:
        print("Resource levels acceptable")
    
    return resource_health

# Test resource management
manage_rebel_resources(50000, 8000, 120)   # All good
manage_rebel_resources(50000, 500, 120)    # Fuel critical
manage_rebel_resources(50000, 8000, 25)    # Food warning
manage_rebel_resources(-1000, 8000, 120)   # Tampered data
```



### Using exception aliases
The **`as`** keyword allows you to assign the **exception** object to a variable, giving you access to the error message and other **exception** attributes:



```python
def validate_coordinates(x, y):
    """Validate coordinate input"""
    try:
        x_coord = float(x)
        y_coord = float(y)
        
        if not (-180 <= x_coord <= 180):
            raise ValueError(f"X coordinate {x_coord} out of range")
            
        return x_coord, y_coord
        
    except ValueError as error_details:
        # Access the specific error message
        print(f"Coordinate error: {error_details}")
        return None, None
```



## Combining functions 
Complex applications are built by composing simple **functions**. Each **function** should have a single, well-defined purpose (Single Responsibility Principle), and multiple **functions** work together to accomplish larger goals.

Real applications chain **functions** together. Each **function** does one thing well, and together they accomplish complex tasks:



```python
def scan_system(system_name):
    """Initial reconnaissance of star system"""
    known_systems = {
        "Coruscant": {"imperial": True, "fleet_size": 1000},
        "Yavin": {"imperial": False, "fleet_size": 45},
        "Hoth": {"imperial": False, "fleet_size": 0}
    }
    return known_systems.get(system_name, {"imperial": None, "fleet_size": 0})

def assess_threat(system_data):
    """Evaluate threat level based on scan data"""
    if system_data["imperial"]:
        if system_data["fleet_size"] > 100:
            return "EXTREME"
        else:
            return "HIGH"
    else:
        return "LOW"

def plan_mission(system_name):
    """Complete mission planning pipeline"""
    try:
        # Step 1: Scan
        intel = scan_system(system_name)
        
        # Step 2: Assess
        threat = assess_threat(intel)
        
        # Step 3: Recommend action
        if threat == "EXTREME":
            action = "Avoid at all costs"
        elif threat == "HIGH":
            action = "Stealth approach only"
        else:
            action = "Safe for standard operations"
            
        return f"System: {system_name} | Threat: {threat} | Recommendation: {action}"
        
    except Exception as e:
        return f"Mission planning failed: {e}"

# Execute complete mission planning
print(plan_mission("Coruscant"))
print(plan_mission("Yavin"))
print(plan_mission("Unknown"))
```



<Note type="note">
    
**Best Practices for Functions**:
- **Single Responsibility**: Each **function** should do one thing well, like a specialized droid
- **Clear Names**: `calculate_jump_distance()` not `calc()` 
- **Documentation**: Add **docstrings** explaining what the **function** does
- **Error Handling**: Always consider what could go wrong
- **Return** Values: Decide if the **function** should **return** data or just perform an action
</Note>

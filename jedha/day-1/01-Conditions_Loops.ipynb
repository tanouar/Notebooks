## Building conditions

You're at the City of Paris reception desk. A 28-year-old resident arrives. Should they go to the youth services office (under 30)? The senior services office (over 65)? Or the general services office? You need to make this decision instantly in your head, conditions teach Python to make the same decisions.

**Conditions** are **decision points** in your code. They **evaluate** whether something is `true` or `false`, then **execute** **different** code based on the **result**. Without conditions, programs can only **perform** the same operations regardless of the data, with conditions, they become **adaptive** and **intelligent**.

### Understanding boolean logic
Before we build conditions, let's understand **Boolean logic**, the foundation of all **decision-making** in programming. Remember from Python Basics that Boolean values are either `True` or `False`. Conditions work by **evaluating** expressions to these **Boolean values**.


````python
# Paul's information
age = 28

# These expressions evaluate to Boolean values
is_adult = age >= 18        # True (28 is greater than or equal to 18)
is_senior = age >= 65        # False (28 is not greater than or equal to 65)
is_youth = age < 30          # True (28 is less than 30)

print(is_adult)   
print(is_senior)  
print(is_youth)   
````


### The if Statement - Your first decision
The `if` statement is the **simplest** condition. It says: "If this is true, do this." The structure is:


```python
if condition:
    # This code runs if condition is True
    # Notice the indentation!
```


Let's check if Paul qualifies for youth programs:


````python
age = 60

if age < 30:
    print("Eligible for youth programs")
    print("Please proceed to Room 101")
````

<Note type="note">
    
**Indentation is CRUCIAL in Python!** The indented lines after the `if` statement form a **code block** that only **executes** when the condition is `True`. Python uses **indentation** to understand which **code** belongs to the condition. Always use 4 spaces (or 1 tab) consistently.
</Note>

### The if/else Statement - Handling both cases
Often, we need to handle both **possibilities** - when something is `true` `AND` when it's `false`. The `else` clause provides an **alternative** path:


````python
age = 75
retirement_age = 65

if age >= retirement_age:
    print("Eligible for senior pension")
    pension = 1200.00
else:
    print("Not yet eligible for pension")
    pension = 0.00

print(f"Monthly pension: €{pension}")
````

The `else` block executes **only when the if condition is False**. It is like: "If this, do that, otherwise do this other thing."

### The elif Statement - Multiple conditions
Real life rarely has just two options. The `elif` (short for "else if") lets us check multiple conditions in sequence:


````python
# Categorizing residents by age group
age = 75

if age < 18:
    category = "Minor"
    services = "Youth Services, Floor 1"
elif age < 30:
    category = "Young Adult"
    services = "Employment Services, Floor 2"
elif age < 65:
    category = "Adult"
    services = "General Services, Floor 3"
else:
    category = "Senior"
    services = "Senior Services, Floor 4"

print(f"Category: {category}")
print(f"Please visit: {services}")
````

`Elif` statement is based on the previous `if` statement, otherwise if the previous conditions are not `True`, we move to the next conditions check. It allows to build many **cases** depending of certains **conditions**. Unlike the `else` statement, it has to check the conditions.  
Python checks each condition **in order** and executes the first True condition it finds, then skips the rest.

### PEP 8 and indentation standards

Python uses indentation (spaces) to define code blocks, and PEP 8 provides clear standards for this. Proper indentation is not just about making code work, it is more about making it professional and maintainable for clear readability.

````python
# GOOD - PEP 8 compliant indentation (4 spaces)
age = 28
if age < 30:
    print("Eligible for youth programs")
    print("Please proceed to Room 101")


# BAD - Inconsistent indentation
if age < 30:
    print("Eligible")  # Only 2 spaces
        print("Room 101")  # Too many spaces
````

### Comparison operators in detail
To build effective conditions, you must master comparison operators. These operators compare values and return Boolean results:

| Operator | Name | Example | Explanation |
|----------|------|---------|-------------|
| `==` | Equal to | `age == 28` | True if age is exactly 28 |
| `!=` | Not equal to | `district != "75001"` | True if district is anything except "75001" |
| `>` | Greater than | `income > 3000` | True if income is strictly greater than 3000 |
| `<` | Less than | `age < 30` | True if age is strictly less than 30 |
| `>=` | Greater or equal | `age >= 18` | True if age is 18 or more |
| `<=` | Less or equal | `income <= 5000` | True if income is 5000 or less |

<Note type="warning">
    
**Common Mistake Alert!**
- `=` is for **assignment**: `age = 28` (stores 28 in age)
- `==` is for **comparison**: `if age == 28:` (checks if age equals 28)
</Note>

### Logical Operators - Combining conditions

**Logical operators** are used to **combine** conditional statements, this allow you to **perform** operations based on multiple **conditions**. These operators are special **keyword** used to carry out **computations** on values and variables.

#### The AND Operator
The Boolean `AND` operator returns `True` if **both** the **conditions** are `True` else it returns `False`.

````python
age = 28
income = 3120.75

# Both conditions must be true
if age >= 18 and income > 2000:
    print("Eligible for standard tax rate")
else:
    print("Special tax consideration required")
````

#### The OR Operator
The Boolean `OR` operator returns `True` if at **least** of the **conditions** is `True`.

````python
age = 28
student_status = False

# Either condition can be true
if age > 26 or student_status == True:
    print("Eligible for youth discount")
else:
    print("Standard pricing applies")
````


#### The NOT Operator
The Boolean `NOT` operator works with a **single** boolean value. If the boolean value is `True` it returns `False` and vice-versa. Otherwise it inverts a Boolean value.

````python
has_parking_permit = True

if not has_parking_permit:
    print("Parking permit required!")
    print("Please visit the permits office")
````

#### Breaking long conditions

When conditions become complex, PEP 8 shows us how to format them clearly, in order to differenciate the conditions.

````python
income = 2500
district_code = "75002"
has_valid_documents = True
process_resident_application = lambda: print("Application processed.")
# GOOD - Breaking long conditions
if (age >= 18 
    and income > 2000
    and district_code in ["75001", "75002", "75003"]
    and has_valid_documents):
        process_resident_application()


# BAD - Unreadable long line
#if age >= 18 and income > 2000 and district_code in ["75001", "75002", "75003"] and has_valid_documents:
#    process_resident_application()
````

### Nested conditions
A nested `if` statement is an `if` statement located **within** another `if` or `else`. It can continue with **multiple** layers, allowing us to evaluate **multiple** conditions **sequentially**. It is particularly useful in scenarios where multiple **criteria** need to be checked **before** taking an action.  

````python
age = 28
income = 3120.75
district_code = 75003

# First check age
if age >= 18:
    print("Adult resident")
    
    # Then check income (only for adults)
    if income < 2500:
        print("  Eligible for housing assistance")
        
        # Then check district (only for eligible adults)
        if district_code >= 75018:
            print("    Priority district - expedited processing")
        else:
            print("    Standard processing time")
    else:
        print("  Income exceeds assistance threshold")
else:
    print("Minor - requires guardian consent")
````


## `for` loops
Now let's look at the second **fundamental** concept in the program: **loop building**. These loops make it possible to **iterate** a certain number of **times** on a program.  
Imagine you need to print resident IDs from 1 to 100. Writing 100 print statements would take **time** and be **tedious** and **error-prone**:

````python
print("Resident ID: 1")
print("Resident ID: 2")
print("Resident ID: 3")
# ... 97 more lines
````

**For loops** automate **repetition**. They execute the **same code** multiple times with **different values**, turning hundreds of lines into just a **few**. They're useful when you know exactly how many times to repeat.

### Understanding iteration
**Iteration** means **repeating** a process. In programming, we iterate to:
- Process multiple items
- Repeat calculations
- Generate sequences
- Validate multiple inputs

The `for` loop is Python's primary iteration tool.

### The [`range()`](https://docs.python.org/3/library/stdtypes.html#range) function
Before diving into for loops, let's understand `range()` is a function that generates **sequences of numbers**.  
`range()` allows to **generate** a series of numbers within a **given range**. Depending on how many **time** you are passing to this function, you can decide where that **series** of numbers will **begin** and **end**.

````python
# range(stop) - generates 0 to stop-1
range(5)        

# range(start, stop) - generates start to stop-1
range(1, 6)     

# range(start, stop, step) - generates with custom step
range(0, 10, 2)

liste_nombre = range(0,10,2)
print(list(liste_nombre))
````

````python
for element in liste_nombre:
    print(element)
````


<Note type="warning">
    
**Important:** `range()` EXCLUDES the stop value!
- `range(5)` gives 0-4, NOT 0-5
- `range(1, 6)` gives 1-5, NOT 1-6
</Note>

### Basic `for` loop structure
The `for` loop **repeats** a code block for each **item** in a sequence. This will allow you to iterate a finite number of times over a program:


```python
for variable in sequence:
    # Code to repeat
    # This is the loop body
```


Let's generate resident IDs:


````python
print("Generating resident IDs:")
for id_number in range(1, 6):
    print(f"Resident ID: {id_number}")
````

### How for loops work 
Let's trace through a loop to understand its execution:


````python
# Calculate squares of numbers
for number in range(1, 4):
    square = number * number
    print(f"{number} squared is {square}")

print("Loop complete!")
````


Execution trace:
1. First iteration: `number = 1`, calculates `1 * 1 = 1`, prints
2. Second iteration: `number = 2`, calculates `2 * 2 = 4`, prints
3. Third iteration: `number = 3`, calculates `3 * 3 = 9`, prints
4. No more values in range, exit loop
5. Continue with code after loop

### Iterating over strings
Strings are **sequences** of **characters**, so we can iterate over them:


````python
district = "Marseille"

print("District code breakdown:")
for character in district:
    print(f"Character: {character}")
````

This loops through the string district, with character representing each character in district one by one.

### Using loop variables in calculations
The loop **variable** can be used in any calculation. Indeed, each time that the **value** in the loop change, it can be used in other **purpose** such as calculation with other **variable**.


````python
# Calculate tax for different income levels
base_income = 2000

print("Income Tax Table:")
print("-" * 30)

for multiplier in range(1, 6):
    income = base_income * multiplier
    
    # Simple tax calculation
    if income < 3000:
        tax = income * 0.10
    elif income < 6000:
        tax = income * 0.15
    else:
        tax = income * 0.20
    
    print(f"Income: €{income:>6} | Tax: €{tax:>7.2f}")
````


### Nested `for` loops
A **nested** loop has one loop **inside** of another. These are used for working with **two dimensions**. When a loop is **nested** inside another loop, the **inner loop** runs many times inside the **outer loop**. In each **iteration** of the outer loop, the inner loop will be **re-started**. The inner loop must **finish** all of its iterations **before** the outer loop can continue to its next iteration.

````python
# Generate apartment numbers for a building

print("Apartment Directory:")
for floor in range(1, 4):  # 3 floors
    print(f"\nFloor {floor}:")
    for room in range(1, 5):  # 4 rooms per floor
        apartment_number = f"{floor}-{room:02d}"
        print(f"  Apartment {apartment_number}")
````

It generates a complete apartment directory for a 3-floor building by using **nested** loops to create **apartment numbers** in the format "**floor-room**" (like 1-01, 1-02, etc.), where it **iterates** through each floor (1-3) and then through each of the 4 rooms per floor.


### Combining `for` loops with conditions
Once we have seen `for` loops and conditions, they are strong tools in order to compute strong and flexible programs:

````python
# Process multiple age values
print("Age Category Analysis")
print("-" * 30)

for age in range(15, 75, 10):  # Test ages: 15, 25, 35, 45, 55, 65
    print(f"\nAge: {age}")
    
    # Categorize each age
    if age < 18:
        print("  Category: Minor")
        print("  Services: Requires guardian")
    elif age < 30:
        print("  Category: Young Adult")
        print("  Services: Career counseling available")
    elif age < 65:
        print("  Category: Adult")
        print("  Services: Standard services")
    else:
        print("  Category: Senior")
        print("  Services: Priority assistance")
````

This code creates an **age categorization** system that loops through six **different ages** (15, 25, 35, 45, 55, 65) and **classifies** each one into **demographic categories** (Minor, Young Adult, Adult, or Senior) with corresponding service recommendations


## `while` loops

`For` loops are perfect when you know how many iterations you need. But what if you don't know? What if you need to:
- Keep asking for input until it's valid
- Process data until a condition changes
- Wait for something to happen
- Search until you find a result

**While loops** continue as long as a **condition** is `True`, making them ideal for **indefinite** iteration.

```python
while condition:
    # Code that repeats while condition is True
    # MUST eventually make condition False!
```


### Basic `while` loop example
Let's validate user input:


````python
# Ensure age is reasonable
age = 120 

while age > 90:
    print(f"Age {age} seems incorrect")
    age = age - 50  # Correct the age
    
print(f"Final age: {age}")
````

### The Infinite loop danger
`While` loops have a **critical risk**: if the condition **never** becomes `False`, the loop runs **forever**!


```python
# WARNING: This would run forever!
count = 1
while count > 0:
     print(count)
     count = count + 1  # Makes condition MORE true, not less!
```

<Note type="warning">
    
**Avoiding Infinite Loops:**
1. Always modify the **condition** variable inside the loop
2. Ensure the modification moves toward making the **condition** `False`
3. Consider adding a **safety** counter
4. If stuck in an **infinite** loop, press `Ctrl+C` to stop it
</Note>

### While vs For: When to use each

| Use FOR when: | Use WHILE when: |
|--------------|-----------------|
| You know the number of iterations | Number of iterations is unknown |
| Processing a sequence | Waiting for a condition |
| Counting operations | Validating input |
| Iterating through ranges | Searching for something |

Comparison example:


````python
# FOR: Print exactly 5 IDs
print("Using FOR loop:")
for id in range(1, 6):
    print(f"ID: {id}")

# WHILE: Print IDs until we reach a limit
print("\nUsing WHILE loop:")
current_id = 1
max_id = 5

while current_id <= max_id:
    print(f"ID: {current_id}")
    current_id = current_id + 1
````


## Loop control statements

Sometimes you need more **control** over loop **execution**:
- Exit early when you find what you're looking for
- Skip certain iterations
- Handle special cases

Python provides **break** and **continue** statements for fine-grained loop control.

### The `break` statement
`break` immediately exits the loop, regardless of the condition:


````python
# Search for a specific value
print("Searching for district 75015...")

for code in range(75001, 75021):
    print(f"Checking: {code}")
    
    if code == 75015:
        print(f"Found district {code}!")
        break  # Exit loop immediately
    
print("Search complete")
````


Without `break`, the loop would continue checking 75016-75020 unnecessarily.

### The `continue` statement
`continue` skips the rest of the current iteration and moves to the next:


````python
# Process only even district codes
print("Processing even-numbered districts only:")

for code in range(75001, 75011):
    if code % 2 != 0:  # If odd number
        continue  # Skip this iteration
    
    print(f"Processing district {code}")
````

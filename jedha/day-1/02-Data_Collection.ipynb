## Lists
Imagine you need to track all residents in the 3rd arrondissement. Creating individual variables like `resident1`, `resident2`, `resident3`... would be **impractical** and **impossible** to maintain.   
Lists are Python's most versatile data collection, perfect for storing **ordered sequences** of **items** (known as values) that might need to change. Unlike individual variables, lists let us manage **multiple related values** under a single name.   
List are very flexible and can be resume by:

- **Mutable**: items can be modified, replaced, or removed
- **Ordered**: elements maintain the order in which they were added
- **Index-based**: items are accessed using their position
- Can store **mixed** data types (integers, strings, booleans, even other lists)
- Can contain **duplicate** items

### Creating and accessing lists
Lists can be created in many ways. You can use **square brackets** ```[]``` or with a **method** ```list()```.  
Let's start by creating a list of Paris arrondissements and their population counts by using the differents methods: 

### 1. Using square brackets

We use square brackets ```[]``` to create a list directly.

````python
arrondissements = ["75001", "75002", "75003", "75004", "75005"]
populations = [16888, 20900, 34115, 27769, 58850]

print(arrondissements)
print(populations)
````

### 2. Using [list()](https://docs.python.org/3/library/functions.html#func-list)

Lists can be created with this function ```list()```. It takes the value inside the brackets and put it in a list.

````python
arrondissements = list(("75001", "75002", "75003", "75004", "75005"))
populations = list((16888, 20900, 34115, 27769, 58850))

print(arrondissements)
print(populations)
````

````python
i = 0

for element in arrondissements:
    print(element, i)
    i += 1
````

````python
list(enumerate(arrondissements))
````

Just like strings have methods, lists have their own set of powerful methods. We access list items using their **index** (position), remembering that Python counting starts at 0.   
So here, in the list arrondissments there are 5 items. In order to access the content of an item, you have to write like this:  
```list[position_item] = value_item```

````python
print(arrondissements[0])  # First district: "75001"
print(populations[2])       # Third district population: 34115
````

<Note type="note">
    
Remember: Python uses **zero-based indexing**. The first item is at index 0, the second at index 1, and so on. This is crucial when working with any sequence in Python!

</Note>

But there are other way to access to a list. For example, we can use a **loop** in order to access **all the items** in a list, it lets us go through each item in the list one by one.

````python
for item in arrondissements:
    print(item)
````

Let's see if we can **mix** data types in a lists. We will retake the arrondissements list, change the district to int and add a string ```Paris 01```.

````python
arrondissements = ["Paris-01",75001, "Paris-02", 75002, "Paris-03", 75003, "Paris-04", 75004, "Paris-05", 75005]
print(arrondissements)
````

### Modifying lists
Lists are **mutable**. It means that we can change their contents after creation. We update the elements by **accessing** them via their **index**.  
This is essential when we want to update resident records, let's change the population number for the 2nd arrondissement:

````python
# Population update for district 75002
populations[1] = 21500
print(populations)  # [16888, 21500, 34115, 27769, 58850]
````

### Adding residents to lists
For our City of Paris database, we need to continuously add new residents. Python provides several methods for this:

#### Using [append()](https://docs.python.org/3/tutorial/datastructures.html) - Adding Single Items
The `.append()` method adds **one item** to the **end of a list**.   
So we have a list of our residents, this method let us add one resident but keep in mind the item is added at the end of the list.

````python
residents = ["Paul Pratt", "Marie Dubois", "Jean Martin"]
residents.append("Sophie Bernard")
residents.append(int(32))
print(residents)
````

#### Using [extend()](https://docs.python.org/3/tutorial/datastructures.html) - Adding Multiple Items
When adding **multiple** residents at **once**, `.extend()` is more efficient.  
It's the same idea as `.append()` but for multiple items and still at the end of the list.

````python
new_residents = ["Lucas Petit", "Emma Moreau"]
residents.extend(new_residents)
print(residents)  
````

<Note type="warning">

Be careful with `.append()` when adding **multiple** items, it will add the entire list as a **single nested item** (it means they are contained in another list):
```python
residents.append(["Alice", "Bob"])  # Creates: [..., ["Alice", "Bob"]]
```
Use `.extend()` or the `+` operator instead for flat addition.

</Note>

#### Using [insert()](https://docs.python.org/3/tutorial/datastructures.html) - Adding an Items at a Specific index
`.insert()` adds an item at a **specific** index in a list.  
Since the other methods add items at the end of the list. `insert()` is more precise if you want to add an item at a specific position. The function work this way, you specify the position then put your value: `list.insert(1,new_value)`

````python
residents.insert(4,"Kevin Georges")
print(residents)
````

### Removing residents from lists
After adding new residents, sometimes we need to remove residents who've moved away. Python offers multiple approaches:

| Method | Purpose | Example |
|--------|---------|---------|
| `del(object)` | Remove by index | `del(residents[2])` |
| `.remove()` | Remove by value | `residents.remove("Paul Pratt")` |
| `.pop()` | Remove and return | `moved = residents.pop(1)` |

Let's handle a resident moving out of Paris, we consider Marie Dubois is moving out.

#### Using [del()](https://docs.python.org/3/tutorial/datastructures.html#the-del-statement) - Removing by index
`del()` allows you to delete an **element** via its **index**. It means that you have to specify the index of the element you want to get rid of.  
For Marie, she is the second resident so has the index 1.

````python
del(residents[1])

print(residents)
````

#### Using [remove()](https://docs.python.org/3/tutorial/datastructures.html) - Removing by value
`.remove()` allows you to delete an **element** via the **value** of the element.  
If we have another resident that is moving out such as `Jean Martin`, we will delete his name from the list like this:

````python
residents.remove("Jean Martin")
print(residents)
````

<Note type="warning">

`remove()` function removes only **one** element at a time. If you have **duplicates** that you want to remove, use a while **loop** as in the following example:

</Note>

````python
resident_marseille = ["Marie Dubois", "Jean Martin", "to_remove", "to_remove"]

while 'to_remove' in resident_marseille:
    resident_marseille.remove('to_remove')
print(resident_marseille)
````

#### Using [pop()](https://docs.python.org/3/tutorial/datastructures.html) - Removing and returning an item	
The problem with these functions is that the item is **deleted** and cannot be **retrieved**. Indeed, it often happens to **separate** an item from a list to use it somewhere else, or even to want to **move** this item in the list. In this case, we will use the `pop()` function.

Paul is finally moving and we need to store his name in a variable:

````python
moving_resident = residents.pop(0)
print(f"{moving_resident} has left Paris")
````

### List comprehensions
When processing resident data, we often need to **transform** entire lists. List comprehensions provide a **Pythonic** way to create new lists based on existing ones. Moreover it is a much more concise way to write a list. 

````python
# Calculate income tax (30%) for all residents
incomes = [3120.75, 2850.00, 4200.50, 3750.25]
taxes = [income * 0.30 for income in incomes]

print(taxes)

# Filter high earners (income > 3500)
high_earners = [income for income in incomes if income > 3500]
print(high_earners)


resultat = []
for income in incomes:
    print(income)
    if income > 3500:
        resultat.append(income)
````

## Slicing - Extracting data subsets
Now that we have lists containing dozens or hundreds of items, we often need just a **portion** of that data. Maybe we want to analyze only the first 10 residents, or compare the last 5 districts.   
**Slicing** gives us a powerful way to **extract** exactly the **subset** we need without writing complex loops or creating new lists manually.  

When analyzing Paris districts, we often need specific portions of our data. **Slicing** extracts subsequences using the syntax `[start:end]`:

- start : starting index (included)
- stop : end index (excluded)

<Note type = "warning">

If you don't put anything as `start`, the program will choose by **default** the **index 0**. Conversely, if you don't put anything as end, the program will understand that it must go to the **last index** in the list

</Note>

Let's slice our arrondissements, we want to focus in the center of Paris and get the thrid to fifth arrondissement.

````python
districts = ["75001", "75002", "75003", "75004", "75005", "75006", "75007"]

central = districts[2:5]
print(central)
````

Let's put nothing at the start, in order to get the first three districts

````python
first = districts[:3]
print(first)
````

This time, we do not specify the end. We want to display the last three districts:

````python
last = districts[4:]
print(last)
````

### Negative index
Python's negative indexing counts from the **end**, making it easy to get **recent** data. It starts from `-1` for the **last element** and goes up to `-n` for the first element.

````python
print(districts[-1])
````

Let's see the last three districts

````python
newest = districts[-3:]
print(newest) 
````

Here, we want to exclude the last two districts:

````python
established = districts[:-2]
print(established) 
````

## Dictionaries
Lists work well when **order** matters, but what if we need to access data by **meaningful names** rather than position? When we retrieve Paul's record, we want to ask for his `'income'` or `'district'`, not remember that income is at index 4. Dictionaries provide this **labeled** access.    
While lists organize data by position, **dictionaries** organize data by meaningful **labels** (keys). This mirrors how the City of Paris actually stores resident information:

### Creating resident records
Instead of remembering that index 0 is the name and index 1 is the age, dictionary can be created by placing a sequence of **elements** within curly `{}` braces, separated by a 'comma'. It uses descriptive keys:

````python
paul_record = {
    'first_name': 'Paul',
    'last_name': 'Pratt',
    'age': 28,
    'district': '75003',
    'income': 3120.75
}

print(paul_record)
````

Unlike lists where we access through index, we need to write the **desired label**: `dict['desired_label']`

````python
# Access data by meaningful names
print(paul_record['first_name'])  # Paul
print(paul_record['district'])    # 75003
````

### Adding and updating resident information
Dictionaries are **mutable**, allowing us to **add** or **update** resident records as their circumstances change. 

Paul adds more information about his email and his children. Moreover he gets a promotion thus we have to update his income and moves to the 11th arrondissement.

In order to add, it is like accessing a record but you write the **new label** and **assign** its value: `dict['added_label] = new_value`. To update it's the same idea, you retake the **desired label** and **reassign** the value.

````python
# Add new information
paul_record['email'] = 'paul.pratt@paris.fr'
paul_record['children'] = 0

# Paul gets a promotion
paul_record['income'] = 3450.00

# Paul moves to a new district
paul_record['district'] = '75011'

print(paul_record)

````

### Deleting resident information
In order to **delete** record, it is the same idea as list, Python offers multiple **approaches**:
| Method | Purpose | Example |
|--------|---------|---------|
| `del(key)` | Remove by key | `del(residents['age'])` |
| `.pop()` | Remove by key and returns its value | `removed = residents.pop('age')` |
| `.popitem()` |  Removes and returns the last key-value pair | `removed = residents.popitem()` |
| `.clear()` | Empties the dictionary | `residents.clear()` |


#### Using [del()](https://docs.python.org/3/tutorial/datastructures.html#the-del-statement) - Removing by index
Like for the list `del()` allows you to delete an **element** via its **labels**.   
For Paul's record, the information of children is not relevant. Let's try to delete this data:

````python
del(paul_record['children'])
print(paul_record)
````

#### Using [popitem()](https://docs.python.org/3/library/collections.html#collections.OrderedDict.popitem) - Removing and returning as key-value pair
Sometimes we need to delete the **last item** that was inserted in a dictionary. `popitem()` method is **remove** and **return** the last **key-value** pair.  
Imagine that we have added a wrong information in our record for Paul. Remember that when we add an infortion it is added at the end of the dictionary.

````python
paul_record['wrong'] = 'false_information'
print(paul_record)
````

````python
paul_record.popitem()
````

````python
print(paul_record)
````

#### Using [pop()](https://docs.python.org/3/library/stdtypes.html#list.pop) - Removing and returning as key-value pair
We can also remove a specific **key-value** pair and return the **value** using the pop().  
Same mistake, we have added a wrong information.

````python
paul_record['number'] = '06 14 12 33'
print(paul_record)
````

````python
paul_record.pop('number')
````

````python
print(paul_record)
````

#### Using [clear()](https://docs.python.org/3/library/stdtypes.html#list.clear) - Removing all the dictionary
`clear()` is used to remove **all items** (key-value pairs) from a dictionary. After calling this method, the dictionary will become **empty** and display none.

```python
print(paul_record.clear())
```

### Managing multiple fields
For the City database, we often need to process all fields. There are multiple ways to iterate through a dictionary, depending if you need key, value or both key-value pairs.  
Python provides methods to iterate through dictionaries:

### 1. Iterate through Value
To **iterate** through all **values** of a dictionary in Python you can use `.values()`. It can be employ a `for` loop, accessing each value **sequentially**. This allows you to process or **display** each individual **value** in the dictionary without explicitly **referencing** the corresponding **keys**.

````python
for val in paul_record.keys():
    print(val)
````

````python
paul_record.keys()
````

### 2. Iterate through keys
In Python, just **looping** through the dictionary provides you its **keys**. You can also iterate keys of a dictionary using built-in `.keys()` method.

````python
for keys in paul_record:
    print(keys)
````

### 3. Iterate through  keys and values
You can use the `items()` function to access both **keys** and **items** at the same time. `items()` method **returns** the view object that contains the **key-value** pair as **tuples**. 

````python
for key, value in paul_record.items():
    print(f"{key}: {value}")
````

````python
paul_record.items()
````

<Note type="note">

Dictionary keys must be **immutable** (strings, numbers, tuples) but values can be any type. This is why we use strings like 'first_name' as keys.

</Note>

## Tuples
Not all data should be **mutable**. The GPS coordinates of the Eiffel Tower will never change, neither will someone's date of birth or the historical population of Paris in 1999. Tuples **protect** this data from accidental **modification**. This immutability makes our code **safer** and signals to other developers that this data is meant to be **constant**.  
Some data shouldn't change once set. **Tuples** are **immutable sequences**, perfect for **fixed data**:


````python
# Latitude and longitude
paris_coords = (48.8566, 2.3522)

# past data is fixed
population_1999 = (1999, 2125246)
population_2009 = (2009, 2234105)
population_2019 = (2019, 2190327)

print(paris_coords)
````

A tuple is created by **storing** all the items inside parentheses `()`, separated by commas. A tuple can have **any** number of **items** and they can be of **different** data types.

<Note type="note">

Tuples can contain elements of various data types, including other tuples, lists, dictionaries and even functions.

</Note>

### Accessing tuple data
In order to access to a tupe date it works like **lists**, but any **modification** attempts will fail:

````python
print(paris_coords[0]) 
````

````python
paris_coords[0] = 48.9  # TypeError: tuple does not support item assignment
````

### Concatenate tuple data
If we want to add more information to residents record. Tuples can be **concatenated** using the `+` operator. This operation **combines** two or more tuples to create a **new tuple**.

````python
residents = ("Paul Pratt", 28, "Marie Dubois", 35 ,"Jean Martin", 42)
district =("75003", "75011", "75015")

records = residents + district
print(records)
````

### Slicing of tuple

Sometimes you need to take a sample of data. Slicing a tuple means **creating** a new tuple from a **subset of elements** of the original tuple.  
The slicing syntax is tuple `[start:stop]`

````python
print(records[2:4])
````

### Deleting a tuple

Since tuples are **immutable**, we can not delete **individual** elements of a tuple. However, we can delete an **entire** tuple.

````python
del(records)
````

<Note type="note">

Printing of Tuple after **deletion** results in an **Error**. Because it delete the **whole** variable.

</Note>

### Iterate over tuples

We have our resident record which is a list of **tuples**, and we want to access each element. The `for` loop will help us to go through the list.

````python
residents_record = [("Paul Pratt", 28), ("Marie Dubois", 35) ,("Jean Martin", 42)]
````

````python
for i in residents_record : # iterating on each tuple
    print('first_name : ' , i[0]) # tuple's first element
    print('age : ' , i[1]) # tuples's second element
````

## Combining collections for real data

Real-world data is rarely simple enough for a **single data** collection. A city database needs **lists** of residents (ordered), **dictionaries** for their details (labeled access), and **tuples** for fixed data (immutable). By **nesting** , these collections, putting dictionaries inside lists, lists inside dictionaries. We can model **complex relationships** and build powerful **data systems** that mirror real-world complexity.

Let's try to build a complex **collections**, we will build a **dictionary** which have a **list** inside a **dictionary**.   
Consider the arrondissement and within they are the information of it and the resident

````python
paris_districts = {
    "75001": {
        "name": "Louvre",
        "population": 16888,
        "residents": ["Paul Pratt", "Marie Dubois"],
        "coordinates": (48.8611, 2.3364)
    },
    "75002": {
        "name": "Bourse",
        "population": 20900,
        "residents": ["Jean Martin", "Sophie Bernard"],
        "coordinates": (48.8679, 2.3432)
    }
}

# Access nested data
print(paris_districts["75001"]["name"])  # Louvre
print(paris_districts["75002"]["residents"][0])  # Paul Pratt
````

<Note type="note">

**Choosing the Right Collection:**  
- **Lists**: When order matters and data might change (resident queue, chronological events)  
- **Dictionaries**: When you need labeled access to data (resident records, configuration settings)  
- **Tuples**: When data should remain constant (coordinates, historical records)

</Note>

## From Documents to Key–Value pairs

You already know **document stores** like MongoDB, which store semi-structured JSON documents indexed by `_id`.

A **Key–Value store** takes that simplicity even further:

> It stores each record as a *pair* — a unique **key** and an associated **value**, nothing more.

Think of it as a giant Python dictionary running at scale:

```python
store = {
    "user:1001": "Luke Skywalker",
    "user:1002": "Leia Organa",
    "user:1003": "Han Solo"
}

print(store["user:1002"])  # 👉 "Leia Organa"
```

<Note tytpe="note" title="Why it is super fast?">

In Key–Value systems, every lookup is an **O(1)** operation — meaning **constant time**, regardless of database size. An **O(1)** operation (pronounced “big-O of one”) means that the time it takes to perform the operation does *not* depend on the size of the dataset.

In other words:

> Whether you have 10 elements or 10 million, the operation takes roughly the same amount of time.

### ⚙️ Example

Imagine a **dictionary in Python** (or a **hash map** in computer science terms):

```python
users = {"luke": "Jedi", "vader": "Sith", "leia": "Rebel"}

print(users["vader"])  # lookup
```

When you access `users["vader"]`, Python uses a hash function to jump directly to where “vader” is stored in memory — it doesn’t scan all keys. That’s an **O(1)** lookup.

### 🧠 In contrast

An **O(n)** operation (linear time) scales with the number of elements (e.g. searching through a list):

```python
names = ["luke", "vader", "leia"]
"vader" in names  # must check one by one → O(n)
```

The more items in the list, the longer it takes.

</Note>

## Key–Value basics

Each entry in a Key–Value store has two parts:

| Term      | Description                                                | Example                                                   |
| --------- | ---------------------------------------------------------- | --------------------------------------------------------- |
| **Key**   | The unique identifier                                      | `"user:hs0lo"`                                            |
| **Value** | The data stored (could be a string, number, blob, or JSON) | `"Han Solo"` or `{"name": "Han Solo", "rank": "Captain"}` |

Unlike document databases:

* There’s **no schema**, **no query language**, and **no secondary indexes** by default.
* You must **know the key** to retrieve the value.

That simplicity is the reason for their speed and scalability.

<Note type="note" title="What is a blob?">

A **blob** (Binary Large Object) is a chunk of binary data — such as an image, audio file, or serialized object — stored directly as the value in a database.  
In a Key–Value store, a blob could be anything from a compressed JSON payload to a cached `.png` file.

</Note>

<Note type="important" title="Trade-off: Simplicity vs Flexibility">

Key–Value stores sacrifice query expressiveness for extreme performance and horizontal scalability. This means that **you can't do joins, filters, or aggregations.**

</Note>

## Why Key–Value stores matter for analytics engineers

In analytics pipelines, Key–Value stores are often used as **fast-access layers** on top of slower systems (like SQL warehouses or document stores).

Here are the most common use cases:

| Use Case               | Description                                        | Example                                     |
| ---------------------- | -------------------------------------------------- | ------------------------------------------- |
| **Caching**            | Store results of expensive SQL or API queries      | Cache `avg_sales_by_region` for dashboards  |
| **Session store**      | Keep user sessions, API tokens, or temporary state | Web analytics sessions, ETL checkpoints     |
| **Real-time counters** | Track metrics that increment constantly            | Page views, active users, API requests      |
| **Feature store**      | Store precomputed ML features                      | `"user:42:last_purchase_amount" = 79.99`    |
| **Stream buffer**      | Act as a lightweight queue or message buffer       | Temporary store before writing to warehouse |

In modern data architectures, Redis is often used alongside:

* PostgreSQL / BigQuery (for persistence)
* Airflow / dbt (for orchestration)
* Streamlit / Dash (for real-time dashboards)

<Note type="tip" title="Analogy with caching layers">

Think of Redis as the **RAM of your data stack** —  
it keeps the hottest, most frequently accessed data **in memory** so your dashboards and pipelines stay fast.

</Note>

## Popular Key–Value systems

| System                | Type         | Key Features                                                     | Common Use                            |
| --------------------- | ------------ | ---------------------------------------------------------------- | ------------------------------------- |
| **Redis**             | In-memory    | Data structures (strings, hashes, lists, sets), TTL, persistence | Caching, counters, queues             |
| **Memcached**         | In-memory    | Simple, volatile cache                                           | Temporary cache layer                 |
| **Amazon DynamoDB**   | Cloud-native | Managed KV + document model, horizontal scaling                  | Scalable app backends, telemetry data |
| **RocksDB / LevelDB** | On-disk      | Embedded KV engine                                               | Underlying engine for big systems     |

Redis is by far the most popular for analytics engineers because it is lightweight, open source, and fast.

## Redis: The Star of Key–Value Stores 🚀

Redis stands for **REmote DIctionary Server**. It behaves just like a Python `dict`, but over the network.

**Example (redis-cli)**

```
SET user:hs0lo "Han Solo"
GET user:hs0lo
```

**Example (Python client)**

```python
import redis
r = redis.Redis(host='localhost', port=6379, decode_responses=True)

r.set("user:hs0lo", "Han Solo")
print(r.get("user:hs0lo"))
```

💡 Again Redis stores data **in memory**, not on disk, which explains its speed.
You can still enable persistence (RDB or AOF mode) for durability.

## When (and When Not) to Use a Key–Value Store

| Use It When ✅                            | Avoid It When 🚫                                        |
| ---------------------------------------- | ------------------------------------------------------- |
| You need **ultra-fast reads/writes**     | You need **complex queries or joins**                   |
| Data access is **key-based**             | You must filter or aggregate by value                   |
| You store **ephemeral or computed data** | You need **durable, auditable history**                 |
| Caching or real-time metrics are needed  | You require **full-text search or multi-field filters** |

<Note type="important" title="Complement, not replacement">
Key–Value stores don’t replace your SQL warehouse or document DB —  
they **complement** them by offloading repetitive or latency-sensitive operations.
</Note>

## Quick Demo 🧑‍💻

Let’s simulate a simple caching workflow with Python:


````python
import time

def slow_query():
    time.sleep(2)
    return 42

key = "result:slow_query"

# Try cache first
cached = r.get(key)

for _ in range(1):
    if cached:
        print("Cached result:", cached)
    else:
        print("\n\nCache miss. Running slow query...")
        result = slow_query()
        r.setex(key, 60, result)  # cache for 60 seconds
        print("Stored result:", result)
````

As you can see:

- ✅ First run → cache miss (takes 2s)
- ✅ Next runs → cached result (instant)
